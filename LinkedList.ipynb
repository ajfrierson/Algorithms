```python
class SinglyNode:

  def __init__(self, val, next = None) -> None:
    self.val = val
    self.next = next

  def __str__(self) -> str:
    return str(self.val)
```

```python
Head = SinglyNode(1)
A = SinglyNode(3)
B = SinglyNode(4)
C = SinglyNode(7)

Head.next = A
A.next = B
B.next = C


print(Head)
```

```python
curr = Head
while curr:
  print(curr)
  curr = curr.next
```

```python
# Display linked list - O(n)

def display(head):
  curr = Head
  elements = []
  while curr:
    elements.append(str(curr.val))
    curr = curr.next
  print("--> ".join(elements))

display(Head)
```

```python
# Search for node value - O(n)

def search(head, val):
  curr = Head
  while curr:
    if val == curr.val:
      return True
    curr = curr.next
  return False

search(Head, 0)
```

```python
# Double Linked List

class DoublyNode:
  def __init__(self, val, next=None, prev=None) -> None:
    self.val = val
    self.next = next
    self.prev = prev

  def __str__(self) -> str:
    return str(self.val)

head = tail = DoublyNode(1)
```

```python
def display(head):
  curr = head
  elements = []
  while curr:
    elements.append(str(curr.val))
    curr = curr.next
    print(" <-> ".join(elements))

display(head)
```

```python
# Insert at beginning - O(1)
def insert_at_beginning(head, tail, val):
  new_node = DoublyNode(val, next=head)
  head.prev = new_node
  return new_node, tail

head, tail = insert_at_beginning(head, tail, 5)
display(head)
```

```python
# Insert at the end O(1)

def insert_at_end(head, tail, val):
  new_node = DoublyNode(val, prev=tail)
  tail.next = new_node
  return head, new_node

head, tail = insert_at_end(head, tail, 4)
display(head)
```
